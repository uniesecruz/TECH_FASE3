# 5. Treinamento e Avaliação de Modelos

```python
from src.model_training import split_data, train_logistic_regression, train_random_forest, save_model
from src.utils import calc_performance_metrics, plot_confusion_matrix
from sklearn.model_selection import KFold

## Divisão dos dados
X_train, X_val, X_test, y_train, y_val, y_test = split_data(X_res, y_res)

# Treinamento dos modelos
kf = KFold(n_splits=5, shuffle=True, random_state=376)
param_grid_lr = {'clf__C': [0.01, 0.1, 1, 10], 'clf__penalty': ['l2'], 'clf__solver': ['lbfgs', 'liblinear']}
param_grid_rf = {'clf__n_estimators': [100, 200], 'clf__max_depth': [None, 10, 20], 'clf__min_samples_split': [2, 5], 'clf__min_samples_leaf': [1, 2]}

# Regressão Logística
grid_lr = train_logistic_regression(X_train, y_train, kf, param_grid_lr)
y_pred_val_lr = grid_lr.best_estimator_.predict(X_val)
calc_performance_metrics("Regressão Logística (Validação)", y_val, y_pred_val_lr)
plot_confusion_matrix("Regressão Logística (Validação)", y_val, y_pred_val_lr)

# Random Forest
grid_rf = train_random_forest(X_train, y_train, kf, param_grid_rf)
y_pred_val_rf = grid_rf.best_estimator_.predict(X_val)
calc_performance_metrics("Random Forest (Validação)", y_val, y_pred_val_rf)
plot_confusion_matrix("Random Forest (Validação)", y_val, y_pred_val_rf)

# Salvar modelos
save_model(grid_lr.best_estimator_, 'modelo_lr.pkl')
save_model(grid_rf.best_estimator_, 'modelo_rf.pkl')
```
